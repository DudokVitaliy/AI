```python
from IPython.display import Markdown, display

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Математика": math,
    "Фізика": physics,
    "Англійська мова": english
}

results = {}
for name, scores in subjects.items():
    results[name] = {
        "avg": sum(scores) / len(scores),
        "min": min(scores),
        "max": max(scores)
    }

best_subject = max(results, key=lambda x: results[x]["avg"])

table_content = ""
for name, stats in results.items():
    table_content += f"| {name} | {stats['avg']:.1f} | {stats['min']} | {stats['max']} |\n"

display(Markdown(f"""
### Порівняльна таблиця результатів


| Предмет | Середній бал | Мін. бал | Макс. бал |
| :--- | :---: | :---: | :---: |
{table_content}

**Предмет з найвищим середнім результатом:** {best_subject} ({results[best_subject]['avg']:.1f})
"""))
```
