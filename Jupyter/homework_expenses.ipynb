```python
from IPython.display import Markdown, display

expenses = [1200, 1500, 800, 600, 1000, 400, 900]

sum_expens = sum(expenses)
ave_expens = sum_expens/len(expenses)
max_expens = max(expenses)
min_expens = min(expenses)
high_spending_days = len([x for x in expenses if x > 1000])

display(Markdown(f"""
| Показник | Значення |
| :--- | :--- |
| Загальна сума витрат | {sum_expens} |
| Середнє значення | {ave_expens} |
| Максимальна витрата | {max_expens} |
| Мінімальна витрата | {min_expens} |
| Кількість днів (витрати > 1000) | {high_spending_days} |
"""))
```
