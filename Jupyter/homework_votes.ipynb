```python
from collections import Counter
from IPython.display import Markdown, display

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]

vote_counts = Counter(votes)

leader = vote_counts.most_common(1)[0][0]
top_3 = vote_counts.most_common(3)

table_rows = "\n".join([f"| {fruit} | {count} |" for fruit, count in vote_counts.items()])

display(Markdown(f"### Результати голосування\n\n| Фрукт | Кількість |\n| :--- | :--- |\n{table_rows}"))
display(Markdown(f"**Фрукт-лідер:** {leader}"))
display(Markdown(f"**Топ-3:** {', '.join([f'{f} ({c})' for f, c in top_3])}"))
```
